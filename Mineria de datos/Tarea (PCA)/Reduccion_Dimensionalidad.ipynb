<p style="font-size:0.93em; color:#555555; margin:0 0 4px; line-height:1.7;">El dataset Telco contiene variables categóricas (tipo de contrato, método de pago, servicios contratados) que el PCA estándar no puede tratar directamente. El <strong>Análisis de Correspondencias Múltiples (ACM)</strong> es la técnica análoga para datos cualitativos: transforma las categorías en un espacio de baja dimensión maximizando la inercia explicada, de forma similar a como el PCA maximiza la varianza. La comparación permite evaluar cuál de los dos métodos captura mejor la estructura latente de los datos para este tipo mixto de variables.</p>
